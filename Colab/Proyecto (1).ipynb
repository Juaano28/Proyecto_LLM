```python
%%writefile medi_bot_bilingue_pro_v02.py
import streamlit as st
import pandas as pd
import os
from datetime import datetime
import csv

# LangChain + Gemini
from langchain_community.vectorstores import Chroma
from langchain_google_genai import ChatGoogleGenerativeAI, GoogleGenerativeAIEmbeddings
from langchain.text_splitter import RecursiveCharacterTextSplitter
from langchain.chains import RetrievalQA
from langchain_core.documents import Document
from langchain.prompts import PromptTemplate
from langchain_core.messages import HumanMessage

# ===================== CONFIG =====================
st.set_page_config(page_title="MediBot Pro v2", page_icon="Brain", layout="wide")
st.title("MediBot Pro v2 - 6 Agentes RAG Especializados")

# API Key
st.sidebar.header("Google AI Studio")
api_key = st.sidebar.text_input("API Key", type="password", help="https://aistudio.google.com/app/apikey")
if not api_key:
    st.warning("Ingresa tu API Key")
    st.stop()

# ===================== CARGAR DATOS CLASIFICADOS =====================
@st.cache_data
def cargar_dataset_clasificado():
    ruta = "/content/dataset_clasificado_PC.csv"
    if not os.path.exists(ruta):
        from google.colab import files
        st.info("Sube tu dataset_final_clasificado.csv")
        uploaded = files.upload()
        ruta = list(uploaded.keys())[0]
    df = pd.read_csv(ruta, on_bad_lines='skip', engine='python')
    st.success(f"Dataset clasificado cargado: {len(df):,} casos")
    return df

df = cargar_dataset_clasificado()

# ===================== MODELOS =====================
@st.cache_resource
def cargar_modelos():
    llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", google_api_key=api_key, temperature=0.3)
    embeddings = GoogleGenerativeAIEmbeddings(model="models/text-embedding-004", google_api_key=api_key)
    return llm, embeddings

llm, embeddings = cargar_modelos()

# ===================== CATEGORÍAS =====================
CATEGORIAS = [
    "Symptom/Diagnosis",
    "Treatment/Medication",
    "Prevention/Health Advice",
    "Test/Interpretation",
    "Emergency/Critical",
    "Post-surgery/Recovery"
]

# ===================== CREAR RAGS POR CATEGORÍA (CON BATCHES) =====================
@st.cache_resource(show_spinner="Creando 6 agentes especializados...")
def crear_rags_por_categoria(_df):
    st.info("Creando 6 bases vectoriales especializadas (por lotes)...")
    progress_bar = st.progress(0)
    rags = {}
    splitter = RecursiveCharacterTextSplitter(chunk_size=800, chunk_overlap=100)

    BATCH_SIZE = 100  # ← Tamaño seguro (ajustable)

    for idx, cat in enumerate(CATEGORIAS):
        df_cat = _df[_df['Categoria'] == cat].copy()
        if len(df_cat) == 0:
            st.warning(f"No hay datos para: {cat}")
            continue

        docs = []
        for _, row in df_cat.iterrows():
            if pd.notna(row.get('Pregunta')) and pd.notna(row.get('Respuesta')):
                text = f"Q: {row['Pregunta']}\nA: {row['Respuesta']}"
                docs.append(Document(page_content=text, metadata={"categoria": cat}))

        split_docs = splitter.split_documents(docs)
        total = len(split_docs)
        st.write(f"**{cat}**: {total:,} chunks → insertando por lotes de {BATCH_SIZE}")

        # Crear colección vacía primero
        collection_name = f"medibot_{cat.lower().replace('/', '_').replace(' ', '_')}"
        db = Chroma(collection_name=collection_name, embedding_function=embeddings)
        db.delete_collection()  # Limpiar si existe
        db = Chroma(collection_name=collection_name, embedding_function=embeddings)

        # Insertar por lotes
        for i in range(0, total, BATCH_SIZE):
            batch = split_docs[i:i+BATCH_SIZE]
            db.add_documents(batch)
            st.write(f"   → Insertados {min(i+BATCH_SIZE, total)}/{total}")

        retriever = db.as_retriever(search_kwargs={"k": 6})

        prompt_template = f"""
You are a highly specialized doctor in **{cat}**.
Answer based ONLY on real patient cases.
Be empathetic, clear, and precise.

Context:
{{context}}

Question: {{question}}
Answer in English:"""

        PROMPT = PromptTemplate(template=prompt_template, input_variables=["context", "question"])

        qa = RetrievalQA.from_chain_type(
            llm=llm,
            chain_type="stuff",
            retriever=retriever,
            return_source_documents=True,
            chain_type_kwargs={"prompt": PROMPT}
        )
        rags[cat] = qa
        progress_bar.progress((idx + 1) / len(CATEGORIAS))

    st.success("6 agentes RAG especializados CREADOS CORRECTAMENTE")
    return rags

rags_especializados = crear_rags_por_categoria(df)

# ===================== TRADUCCIÓN Y CLASIFICACIÓN =====================
def traducir(texto, destino="en"):
    prompt = f"Translate to {'English' if destino=='en' else 'Spanish'} (only the text):\n\n{texto}"
    try:
        return llm.invoke([HumanMessage(content=prompt)]).content.strip()
    except:
        return texto

def clasificar_pregunta(pregunta_en):
    prompt = f"""
Classify into EXACTLY ONE category:
{', '.join(CATEGORIAS)}

Question: {pregunta_en}

Answer ONLY the category name."""
    try:
        resp = llm.invoke([HumanMessage(content=prompt)]).content.strip()
        resp = resp.replace('"', '').replace("'", "").strip()
        return resp if resp in CATEGORIAS else "Symptom/Diagnosis"
    except:
        return "Symptom/Diagnosis"

# ===================== CHATBOT =====================
st.subheader("Chatbot con 6 Especialistas Médicos")

if "messages" not in st.session_state:
    st.session_state.messages = [{"role": "assistant", "content": "¡Hola! Tengo 6 especialistas listos. Pregúntame en español."}]

for msg in st.session_state.messages:
    with st.chat_message(msg["role"]):
        st.write(msg["content"])

if prompt := st.chat_input("¿En qué te ayudo hoy?"):
    st.session_state.messages.append({"role": "user", "content": prompt})
    with st.chat_message("user"):
        st.write(prompt)

    with st.chat_message("assistant"):
        with st.spinner("Traduciendo..."):
            pregunta_en = traducir(prompt, "en")

        with st.spinner("Asignando especialista..."):
            categoria = clasificar_pregunta(pregunta_en)
            st.info(f"Especialista: **{categoria}**")

        if categoria == "Emergency/Critical":
            respuesta = "¡EMERGENCIA!\n\nVe inmediatamente a urgencias o llama al número de emergencias de tu país."
            st.error(respuesta)
        else:
            with st.spinner(f"Consultando al especialista en {categoria}..."):
                qa = rags_especializados.get(categoria)
                if qa:
                    result = qa({"query": pregunta_en})
                    respuesta_en = result["result"]
                    fuentes = result["source_documents"]
                else:
                    respuesta_en = "Lo siento, aún no tengo suficientes casos en esta categoría."

            respuesta_es = traducir(respuesta_en, "es")
            st.write(respuesta_es)

            if 'fuentes' in locals():
                with st.expander(f"Fuentes reales ({len(fuentes)})"):
                    for doc in fuentes[:3]:
                        st.caption(doc.page_content[:600] + "...")

        st.session_state.messages.append({"role": "assistant", "content": respuesta_es})
```
