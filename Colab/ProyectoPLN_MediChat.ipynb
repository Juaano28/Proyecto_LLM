```python
import pandas as pd
import json

# === 1. Cargar datasets ===
train_mcc = pd.read_csv('train.csv')
test_mcc = pd.read_csv('test.csv')

with open('english-train.json', 'r') as f:
    meddialog_data = json.load(f)

mts_train = pd.read_csv('MTS-Dialog-TrainingSet.csv')
mts_val = pd.read_csv('MTS-Dialog-ValidationSet.csv')
mts_test1 = pd.read_csv('MTS-Dialog-TestSet-1-MEDIQA-Chat-2023.csv')
mts_test2 = pd.read_csv('MTS-Dialog-TestSet-2-MEDIQA-Sum-2023.csv')

# === 2. Función de unificación ===
def unify(df, source, text_col=None, summary_col=None):
    df = df.copy()
    df['source'] = source
    df['conversation'] = df[text_col].astype(str) if text_col in df.columns else None
    df['summary'] = df[summary_col].astype(str) if summary_col in df.columns else None
    return df[['source', 'conversation', 'summary']]

# === 3. Adaptar estructura según dataset ===

# Medical Conversation Corpus
train_mcc_u = unify(train_mcc, "MedicalCorpus_train", "Conversation")
test_mcc_u = unify(test_mcc, "MedicalCorpus_test", "Conversation")

# MedDialog (utterances es lista → concatenamos)
meddialog_df = pd.DataFrame(meddialog_data)
meddialog_df["dialogue"] = meddialog_df["utterances"].apply(lambda utts: " ".join(utts) if isinstance(utts, list) else str(utts))
meddialog_df["summary"] = meddialog_df["description"]
meddialog_u = unify(meddialog_df, "MedDialog", "dialogue", "summary")

# MTS Dialog
mts_train_u = unify(mts_train, "MTS_Training", "dialogue", "section_text")
mts_val_u = unify(mts_val, "MTS_Validation", "dialogue", "section_text")
mts_test1_u = unify(mts_test1, "MTS_Test1", "dialogue", "section_text")
mts_test2_u = unify(mts_test2, "MTS_Test2", "dialogue", "section_text")

# === 4. Unir todo ===
unified = pd.concat([
    train_mcc_u, test_mcc_u, meddialog_u,
    mts_train_u, mts_val_u, mts_test1_u, mts_test2_u
], ignore_index=True)

# === 5. Mostrar información ===
print("✅ Dataset unificado correctamente")
print(f"Total de registros combinados: {len(unified)}\n")
print("Columnas:", unified.columns.tolist())

# === 6. Mostrar ejemplo de cada dataset ===
print("\n=== Ejemplo de diálogos ===")
for source in unified['source'].unique():
    sample = unified[unified['source'] == source].dropna(subset=['conversation']).head(1)
    if not sample.empty:
        print(f"\n🔹 {source}:")
        print(sample['conversation'].values[0][:400], "...")  # Mostrar primeros 400 caracteres

# === 7. Guardar CSV ===
unified.to_csv('Unified_Medical_Dialogs.csv', index=False)
print("\nArchivo final guardado como 'Unified_Medical_Dialogs.csv'")
```

```python
# ==============================================
# 🩺 MediChat - Unificación de los 7 datasets médicos
# ==============================================

import pandas as pd
import json
import re

# ---------- Función de limpieza ----------
def limpiar_texto(texto):
    if pd.isna(texto):
        return ""
    texto = re.sub(r"http\S+", "", texto)
    texto = re.sub(r"[^a-zA-ZáéíóúüñÁÉÍÓÚÜÑ0-9\s,.?!]", " ", texto)
    texto = re.sub(r"\s+", " ", texto)
    return texto.strip().lower()

# ---------- Dataset 1: english-train.json ----------
json_rows = []
try:
    with open("english-train.json", "r") as f:
        data_json = json.load(f)
    for i, d in enumerate(data_json):
        utterances = d.get("utterances", [])
        patient = next((u.replace("Patient:", "").replace("patient:", "").strip()
                        for u in utterances if u.lower().startswith("patient:")), "")
        doctor = next((u.replace("Doctor:", "").replace("doctor:", "").strip()
                       for u in utterances if u.lower().startswith("doctor:")), "")
        if patient and doctor:
            json_rows.append({
                "source": "MedDialog",
                "idioma": "en",
                "pregunta": limpiar_texto(patient),
                "respuesta": limpiar_texto(doctor)
            })
    df_json = pd.DataFrame(json_rows)
    print(f"✅ Cargado english-train.json → {df_json.shape}")
except Exception as e:
    print("⚠️ Error cargando english-train.json:", e)
    df_json = pd.DataFrame(columns=["source","idioma","pregunta","respuesta"])

# ---------- Dataset 2: train.csv y test.csv ----------
def extraer_conversaciones(nombre_archivo):
    filas = []
    try:
        df = pd.read_csv(nombre_archivo)
        for text in df["Conversation"]:
            if not isinstance(text, str):
                continue
            # Buscar texto entre [|Human|] y [|AI|]
            partes = re.findall(r"\[\|Human\|\](.*?)\[\|AI\|\](.*)", text, flags=re.S)
            for human, ai in partes:
                if human.strip() and ai.strip():
                    filas.append({
                        "source": "MedicalConversationCorpus",
                        "idioma": "en",
                        "pregunta": limpiar_texto(human),
                        "respuesta": limpiar_texto(ai)
                    })
        df_out = pd.DataFrame(filas)
        print(f"✅ Cargado {nombre_archivo} → {df_out.shape}")
        return df_out
    except Exception as e:
        print(f"⚠️ Error en {nombre_archivo}: {e}")
        return pd.DataFrame(columns=["source","idioma","pregunta","respuesta"])

df_train_mcc = extraer_conversaciones("train.csv")
df_test_mcc = extraer_conversaciones("test.csv")

# ---------- Datasets MTS-Dialog ----------
def cargar_mts(nombre_archivo):
    try:
        df = pd.read_csv(nombre_archivo)
        df = df.rename(columns={
            "dialogue": "pregunta",
            "section_text": "respuesta"
        })
        df["source"] = "MTS-Dialog"
        df["idioma"] = "en"
        df["pregunta"] = df["pregunta"].apply(limpiar_texto)
        df["respuesta"] = df["respuesta"].apply(limpiar_texto)
        print(f"✅ Cargado {nombre_archivo} → {df.shape}")
        return df[["source","idioma","pregunta","respuesta"]]
    except Exception as e:
        print(f"⚠️ Error cargando {nombre_archivo}: {e}")
        return pd.DataFrame(columns=["source","idioma","pregunta","respuesta"])

mts_train = cargar_mts("MTS-Dialog-TrainingSet.csv")
mts_val = cargar_mts("MTS-Dialog-ValidationSet.csv")
mts_test1 = cargar_mts("MTS-Dialog-TestSet-1-MEDIQA-Chat-2023.csv")
mts_test2 = cargar_mts("MTS-Dialog-TestSet-2-MEDIQA-Sum-2023.csv")

# ---------- Concatenar todos ----------
df_total = pd.concat([
    df_json, df_train_mcc, df_test_mcc,
    mts_train, mts_val, mts_test1, mts_test2
], ignore_index=True)

# ---------- Limpieza final ----------
df_total = df_total.dropna(subset=["pregunta","respuesta"])
df_total = df_total.drop_duplicates(subset=["pregunta"])
print(f"\n📊 Dataset final unificado: {df_total.shape[0]} registros totales")

print("\nEjemplo de registros unificados:")
print(df_total.sample(5, random_state=42))

# ---------- Guardar archivo final ----------
df_total.to_csv("medichat_dataset_final.csv", index=False)
print("\n💾 Archivo guardado como: medichat_dataset_final.csv")
```

```python
# ==============================================
# 🌍 MediChat - Traducción inglés → español (para dataset bilingüe)
# ==============================================

from transformers import MarianMTModel, MarianTokenizer
import pandas as pd
import torch
from tqdm import tqdm

# Cargar dataset unificado
df = pd.read_csv("medichat_dataset_final.csv")

# Cargar modelo de traducción Helsinki-NLP (en→es)
model_name = "Helsinki-NLP/opus-mt-en-es"
tokenizer = MarianTokenizer.from_pretrained(model_name)
model = MarianMTModel.from_pretrained(model_name)

device = "cuda" if torch.cuda.is_available() else "cpu"
model = model.to(device)

def traducir_texto(texto):
    if not isinstance(texto, str) or texto.strip() == "":
        return ""
    batch = tokenizer([texto], return_tensors="pt", truncation=True, padding=True).to(device)
    translated = model.generate(**batch, max_length=256)
    return tokenizer.decode(translated[0], skip_special_tokens=True)

# Traducir solo una muestra al principio (para probar)
df_sample = df.sample(50, random_state=1).copy()

tqdm.pandas()
df_sample["pregunta_es"] = df_sample["pregunta"].progress_apply(traducir_texto)
df_sample["respuesta_es"] = df_sample["respuesta"].progress_apply(traducir_texto)

# Mostrar algunas traducciones
print("\n✅ Ejemplo de traducción:")
print(df_sample[["pregunta", "pregunta_es", "respuesta_es"]].head(5))

# Guardar dataset bilingüe (muestra)
df_sample.to_csv("medichat_dataset_bilingue_muestra.csv", index=False)
print("\n💾 Archivo guardado: medichat_dataset_bilingue_muestra.csv")
```

```python
batch_size = 16
preguntas_es, respuestas_es = [], []

for i in tqdm(range(0, len(df), batch_size)):
    batch_q = df["pregunta"].iloc[i:i+batch_size].tolist()
    batch_a = df["respuesta"].iloc[i:i+batch_size].tolist()

    # Traducción preguntas
    tokens_q = tokenizer(batch_q, return_tensors="pt", padding=True, truncation=True).to(device)
    out_q = model.generate(**tokens_q, max_length=256)
    preguntas_es.extend([tokenizer.decode(t, skip_special_tokens=True) for t in out_q])

    # Traducción respuestas
    tokens_a = tokenizer(batch_a, return_tensors="pt", padding=True, truncation=True).to(device)
    out_a = model.generate(**tokens_a, max_length=256)
    respuestas_es.extend([tokenizer.decode(t, skip_special_tokens=True) for t in out_a])

df["pregunta_es"] = preguntas_es
df["respuesta_es"] = respuestas_es
df.to_csv("medichat_dataset_bilingue.csv", index=False)
print("💾 Dataset bilingüe completo guardado")
```
